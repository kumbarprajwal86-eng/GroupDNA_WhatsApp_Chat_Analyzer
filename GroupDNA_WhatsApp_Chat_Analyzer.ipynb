```python
import numpy as np
from datetime import datetime, timedelta

print("GroupDNA Project Started Successfully!")
```

```python
file_path = "/content/hostel_bois.txt"

with open(file_path, "r", encoding="utf-8") as file:
    content = file.read()

print("Dataset loaded successfully!")
print("Total characters:", len(content))
```

```python
lines = content.split("\n")

print("Total lines:", len(lines))

print("\nFirst 5 lines of the dataset:")

for i in range(5):
    print(lines[i])
```

```python
def is_date_line(line):

    if len(line) < 8:
        return False

    if (line[0:2].isdigit() and
        line[2] == "/" and
        line[3:5].isdigit() and
        line[5] == "/" and
        line[6:8].isdigit()):

        return True

    return False
```

```python
messages = []

system_count = 0
media_count = 0
deleted_count = 0

media_per_person = {}
deleted_per_person = {}

participants = set()

# Combine multi-line messages
combined_lines = []
current_line = ""

for line in lines:

    line = line.rstrip("\n")

    if line.strip() == "":
        continue

    if is_date_line(line):

        if current_line != "":
            combined_lines.append(current_line)

        current_line = line.strip()

    else:

        if current_line != "":
            current_line = current_line + " " + line.strip()

if current_line != "":
    combined_lines.append(current_line)


# Extract timestamp, sender and message
for line in combined_lines:

    if " - " not in line:
        system_count += 1
        continue

    parts = line.split(" - ", 1)

    timestamp = parts[0]
    remaining = parts[1]

    if ": " not in remaining:
        system_count += 1
        continue

    sender_parts = remaining.split(": ", 1)

    sender = sender_parts[0]
    message_text = sender_parts[1]

    try:
        message_date = datetime.strptime(
            timestamp, "%d/%m/%y, %H:%M"
        )
    except ValueError:
        system_count += 1
        continue

    message_type = "normal"

    if message_text == "<Media omitted>":

        message_type = "media"
        media_count += 1

        media_per_person[sender] = (
            media_per_person.get(sender, 0) + 1
        )

    elif message_text == "This message was deleted":

        message_type = "deleted"
        deleted_count += 1

        deleted_per_person[sender] = (
            deleted_per_person.get(sender, 0) + 1
        )

    participants.add(sender)

    message_data = {
        "timestamp": timestamp,
        "datetime": message_date,
        "sender": sender,
        "text": message_text,
        "type": message_type
    }

    messages.append(message_data)


print("CHAT PARSING COMPLETED!")
print("--------------------------------")

print("Total parsed messages:", len(messages))
print("Total participants:", len(participants))
print("System messages:", system_count)
print("Media messages:", media_count)
print("Deleted messages:", deleted_count)

normal_count = len(messages) - media_count - deleted_count

print("Normal text messages:", normal_count)
```

```python
participants = []

for msg in messages:
    if msg["sender"] != "System":
        if msg["sender"] not in participants:
            participants.append(msg["sender"])

print("Total Participants:", len(participants))
print("Participants:", participants)
```

```python
message_counts = {}

for msg in messages:
    sender = msg["sender"]

    if sender != "System":
        if sender not in message_counts:
            message_counts[sender] = 1
        else:
            message_counts[sender] += 1

print("Messages sent by each participant:")

for sender, count in message_counts.items():
    print(sender, ":", count)
```

```python
most_active = max(message_counts, key=message_counts.get)

print("Most Active Participant:", most_active)
print("Messages Sent:", message_counts[most_active])
```

```python
date_counts = {}

for msg in messages:
    if msg["sender"] != "System":
        date = msg["datetime"].date()

        if date not in date_counts:
            date_counts[date] = 1
        else:
            date_counts[date] += 1

busiest_date = max(date_counts, key=date_counts.get)

print("Busiest Date:", busiest_date)
print("Messages:", date_counts[busiest_date])
```

```python
weekday_counts = {}

for msg in messages:
    if msg["sender"] != "System":
        weekday = msg["datetime"].strftime("%A")

        if weekday not in weekday_counts:
            weekday_counts[weekday] = 1
        else:
            weekday_counts[weekday] += 1

busiest_weekday = max(weekday_counts, key=weekday_counts.get)

print("Busiest Weekday:", busiest_weekday)
print("Messages:", weekday_counts[busiest_weekday])
```

```python
hour_counts = {}

for msg in messages:
    if msg["sender"] != "System":
        hour = msg["datetime"].hour

        if hour not in hour_counts:
            hour_counts[hour] = 1
        else:
            hour_counts[hour] += 1

busiest_hour = max(hour_counts, key=hour_counts.get)

print("Busiest Hour:", busiest_hour)
print("Messages:", hour_counts[busiest_hour])
```

```python
import numpy as np

activity_matrix = np.zeros((7, 24), dtype=int)

for msg in messages:
    if msg["sender"] != "System":
        weekday = msg["datetime"].weekday()
        hour = msg["datetime"].hour

        activity_matrix[weekday][hour] += 1

print("NumPy Activity Matrix:")
print(activity_matrix)
```

```python
weekdays = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
]

print("Activity Heatmap (Messages by Weekday and Hour)")

print("Day       ", end="")

for hour in range(24):
    print(f"{hour:3}", end="")

print()

for i in range(7):
    print(f"{weekdays[i]:10}", end="")

    for j in range(24):
        print(f"{activity_matrix[i][j]:3}", end="")

    print()
```

```python
import string
word_counts = {}

stop_words = [
    "the", "is", "and", "to", "in", "of", "a", "an",
    "you", "i", "it", "for", "on", "this", "that",
    "are", "was", "we", "he", "she", "they", "me",
    "my", "your", "have", "has", "had", "be", "at",
    "with", "what", "but", "so", "if", "not", "do",
    "will", "can", "just", "yes", "no", "ok"
]

for msg in messages:
    if msg["sender"] != "System":

        message_text = msg["text"].lower()

        for symbol in string.punctuation:
            message_text = message_text.replace(symbol, " ")

        words = message_text.split()

        for word in words:
            if word not in stop_words and len(word) > 2:

                if word not in word_counts:
                    word_counts[word] = 1
                else:
                    word_counts[word] += 1

print("Total Unique Words:", len(word_counts))
```

```python
top_words = sorted(
    word_counts.items(),
    key=lambda item: item[1],
    reverse=True
)

print("Top 10 Most Used Words:")

for word, count in top_words[:10]:
    print(word, ":", count)
```

```python
response_times = {}

for i in range(1, len(messages)):
    previous = messages[i - 1]
    current = messages[i]

    previous_sender = previous["sender"]
    current_sender = current["sender"]

    if previous_sender != "System" and current_sender != "System":
        if previous_sender != current_sender:

            time_difference = (
                current["datetime"] - previous["datetime"]
            ).total_seconds() / 60

            if current_sender not in response_times:
                response_times[current_sender] = []

            response_times[current_sender].append(time_difference)

print("Average Response Time:")

for sender, times in response_times.items():
    average = sum(times) / len(times)
    print(sender, ":", round(average, 2), "minutes")
```

```python
average_response_time = {}
for sender, times in response_times.items():
    average_response_time[sender] = sum(times) / len(times) if times else 0

# We don't have a precalculated longest_silent_streak, so we default to 0 for analysis
longest_silent_streak = {sender: 0 for sender in participants}

personality = {}

for sender in participants:
    count = message_counts.get(sender, 0)
    response = average_response_time.get(sender, 999)
    silence = longest_silent_streak.get(sender, 0)

    if count >= 500 and response <= 40:
        personality[sender] = "Active Chatter"
    elif response <= 35:
        personality[sender] = "Quick Responder"
    elif count <= 100:
        personality[sender] = "Silent Observer"
    elif silence >= 100:
        personality[sender] = "Silent Observer"
    else:
        personality[sender] = "Regular Member"

print("Group Personality Analysis:")
for sender, category in personality.items():
    print(sender, ":", category)
```

```python
print("\nGROUPDNA - PERSONALITY REPORT")
print("-" * 40)

for sender in participants:
    print("Participant:", sender)
    print("Messages:", message_counts[sender])
    print("Personality:", personality[sender])
    print("-" * 40)
```

```python
from datetime import timedelta

longest_silent_streak = {}

for sender in participants:

    user_messages = []

    for msg in messages:
        if msg["sender"] == sender:
            user_messages.append(msg["datetime"])

    user_messages.sort()

    max_gap = timedelta(seconds=0)

    for i in range(1, len(user_messages)):
        gap = user_messages[i] - user_messages[i - 1]

        if gap > max_gap:
            max_gap = gap

    longest_silent_streak[sender] = round(
        max_gap.total_seconds() / 3600, 2
    )

print("Corrected Longest Silent Streaks:")

for sender, hours in longest_silent_streak.items():
    print(sender, ":", hours, "hours")
```
